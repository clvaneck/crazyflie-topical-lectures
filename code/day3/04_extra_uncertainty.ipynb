# Optional: uncertainty and speed

This simulation connects Day 2's state-estimation uncertainty to gate clearance and a recommended speed. Choose nominal measurements, a dropout, an outlier or both, and predict the effect.

Edit the policy below: choose the clearance at which cruise speed becomes appropriate, or change how speed increases with clearance. The helper calculates clearance after accounting for the drone, tracking allowance and projected uncertainty. It rejects non-finite or out-of-range speeds and requires a stop when clearance is non-positive.

Compare the recommendation with the uncertainty and clearance plots. This remains an optional extension: the recommended speed does not change the prescribed motion, so elapsed simulation time cannot predict a race result.


```python
from pathlib import Path
import sys
root = next(path for path in (Path.cwd(), *Path.cwd().parents)
            if (path / "code/day3/plan.py").is_file())
sys.path.insert(0, str(root / "code"))
from IPython.display import display
from day3.uncertainty import run, plot
```

```python
def our_speed_policy(clearance_m, nominal_speed_m_s):
    if clearance_m <= 0:
        return 0.0
    clearance_for_cruise_m = 0.15
    speed_fraction = min(1.0, clearance_m / clearance_for_cruise_m)
    return nominal_speed_m_s * speed_fraction
```

```python
scenario = "dropout"  # "nominal", "outlier", "combined"
seed = 7
records = run(seed, scenario, speed_policy=our_speed_policy)
figure = plot(records, scenario, show=False)
display(figure)
```
